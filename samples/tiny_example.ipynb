# Churn features
Builds weekly churn features.

```python
events = spark.table("raw.events")
WINDOW_DAYS = 28
recent = events.filter(f"event_date >= date_sub(current_date(), {WINDOW_DAYS})")
```

```python
%sql
SELECT customer_id, COUNT(*) AS n FROM raw.events GROUP BY customer_id
```

```python
WINDOW_DAYS = 7
recent.groupBy("customer_id").count().write.mode("append").saveAsTable("features.churn_weekly")
```
